```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import convolve2d
from skimage import data, color
```

**step 2 applying filter**
detects top edges

```python
# Load grayscale image
image = color.rgb2gray(data.coffee())

# Define a simple edge detection filter
filter = np.array([[-1, -1, -1],
                   [ 0,  0,  0],
                   [ 1,  1,  1]])
```

```python



# Apply convolution
convolved = convolve2d(image, filter, mode='same', boundary='symm')

# Show result
plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.title("Original")
plt.imshow(image, cmap='gray')
plt.subplot(1, 2, 2)
plt.title("After Convolution")
plt.imshow(convolved, cmap='gray')
plt.show()
```

**detetecting bottom edges**

```python
filter2 = np.array([[1, 1, 1],
                   [ 0,  2, 0],
                   [ -1, - 1,-1]])

# Apply convolution
convolved = convolve2d(image, filter2, mode='same', boundary='symm')

# Show result
plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.title("Original")
plt.imshow(image, cmap='gray')
plt.subplot(1, 2, 2)
plt.title("After Convolution")
plt.imshow(convolved, cmap='gray')
plt.show()
```

```python
filter3 = np.array([[-1, 0, 1],
                   [ -1,  0, 1],
                   [ -1, 0,1]])

# Apply convolution
convolved = convolve2d(image, filter3, mode='same', boundary='symm')

# Show result
plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.title("Original")
plt.imshow(image, cmap='gray')
plt.subplot(1, 2, 2)
plt.title("right edges")
plt.imshow(convolved, cmap='gray')
plt.show()
```

```python
filter4 = np.array([[1, 0, -1],
                   [ 1, 0,-1],
                   [ 1, 0,-1]])

# Apply convolution
convolved = convolve2d(image, filter4, mode='same', boundary='symm')

# Show result
plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.title("Original")
plt.imshow(image, cmap='gray')
plt.subplot(1, 2, 2)
plt.title("left edges")
plt.imshow(convolved, cmap='gray')
plt.show()
```
